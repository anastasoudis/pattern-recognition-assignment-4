**Ερώτημα Α.**

```python
from ucimlrepo import fetch_ucirepo
from sklearn.model_selection import train_test_split

# Φόρτωση δεδομένων
breast_cancer_wisconsin_diagnostic = fetch_ucirepo(id=17)

X = breast_cancer_wisconsin_diagnostic.data.features
y = breast_cancer_wisconsin_diagnostic.data.targets

# Το 70% των δεδομένων θα χρησιμοποιηθεί για την εκπαίδευση και το 30% για τη δοκιμή
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Εκτύπωση των διαστάσεων
print("Training set dimensions:", X_train.shape)
print("Test set dimensions:", X_test.shape)
```

**Ερώτημα Β.**

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score
from sklearn.impute import SimpleImputer
import pandas as pd
import numpy as np

# Χρήση SimpleImputer για τα missing values
imputer = SimpleImputer(strategy='mean')
X_train_imputed = imputer.fit_transform(X_train)

# Μετατροπή του αποτελέσματος σε DataFrame και διατήρηση των ονομάτων χαρακτηριστικών
X_train_imputed = pd.DataFrame(X_train_imputed, columns=X_train.columns)

# Εκπαίδευση δενδρικού ταξινομητή
clf = DecisionTreeClassifier(max_depth=5)
clf.fit(X_train_imputed, y_train)

# Υπολογισμός της ακρίβειας
y_pred = clf.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)
print("Accuracy:", accuracy)
```
